## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """
    Validate a sequencing read against a reference sequence.

    Checks that the read is the same length as the reference and contains
    only valid DNA characters (A, C, G, T). Lowercase letters are converted
    to uppercase. Returns a dictionary with the uppercased sequence, a
    validity flag, and a list of any errors found.

    Args:
        sequence (str): The read sequence to validate (any length/content).
        reference (str): The reference sequence (uppercase A/C/G/T only).

    Returns:
        dict: Keys are 'sequence' (str, uppercased), 'valid' (bool),
              and 'errors' (list of str).
    """
    # Convert sequence to uppercase (repair lowercase reads)
    sequence = sequence.upper()

    # Start with an empty list of errors
    errors = []

    # Checking if the length matches the reference
    if len(sequence) != len(reference):
        errors.append("wrong_length")

    # Checking if all characters are valid DNA bases (A, C, G, T)
    valid_bases = set("ACGT")
    if not all(base in valid_bases for base in sequence):
        errors.append("invalid_characters")

    # Read is valid only if there are no errors
    is_valid = len(errors) == 0

    # Return the result as a dictionary
    return {
        "sequence": sequence,
        "valid": is_valid,
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence, reference):
    """
    Find all mismatches between a sequencing read and the reference sequence.

    Compares the read and reference base by base, returning a list of
    dictionaries describing each mismatch. Positions are numbered from 1.
    Returns an empty list if the read perfectly matches the reference.

    Args:
        sequence (str): The read sequence (valid, uppercase A/C/G/T, same length as reference).
        reference (str): The reference sequence (valid, uppercase A/C/G/T).

    Returns:
        list: A list of dictionaries, each with keys 'position' (int, 1-indexed),
              'reference_base' (str), and 'read_base' (str). Empty list if no mismatches.
    """
    # Start with an empty list to store mutations
    mutations = []

    # Loop through each position in the sequence
    for i in range(len(sequence)):
        # Compare the base in the read to the base in the reference
        if sequence[i] != reference[i]:
            # Found a mismatch — record it
            mutation = {
                "position": i + 1,              # +1 because positions start at 1, not 0
                "reference_base": reference[i], # What the reference has here
                "read_base": sequence[i]        # What the read has here
            }
            mutations.append(mutation)

    return mutations
````

````python
def analyse_reads(reads, reference):
    """Analyse all reads by validating and finding mutations.

    Uses validate_read and find_mutations to process each read.
    Returns a new list of dicts (does not modify the original reads).

    Args:
        reads: List of dicts with keys 'read_id', 'sample', 'group', 'sequence'.
        reference: A valid uppercase nucleotide string.

    Returns:
        A list of dicts, one per read, with keys:
            'read_id', 'sample', 'group', 'sequence' (uppercase),
            'status' ('valid' or 'invalid'), 'errors' (list),
            'mutations' (list of mismatch dicts, or [] if invalid).
        Returns [] if reads is empty.
    """
    analysed = []

    for read in reads:
        # Validate the read
        validation = validate_read(read["sequence"], reference)

        # Build the output dictionary
        result = {
            "read_id": read["read_id"],
            "sample": read["sample"],
            "group": read["group"],
            "sequence": validation["sequence"],  # uppercase version
            "status": "valid" if validation["valid"] else "invalid",
            "errors": validation["errors"],
            "mutations": [],  # default to empty
        }

        # Only find mutations if the read is valid
        if validation["valid"]:
            result["mutations"] = find_mutations(validation["sequence"], reference)

        analysed.append(result)

    return analysed
````

### 💻 Enter your code here

````python
def summarise_samples(analysed_reads):
    """Summarise mismatch patterns per sample from analysed reads.

    Args:
        analysed_reads: List of dicts as returned by analyse_reads().

    Returns:
        A nested dict keyed by sample name. Each value is a dict with:
            'group', 'valid_reads', 'invalid_reads', 'mismatch_reads',
            'mismatch_percentage', 'mismatch_events', 'substitution_counts',
            'position_counts', 'top_positions'.
        Returns {} for empty input.
    """
    if not analysed_reads:
        return {}

    # Group reads by sample
    samples = {}
    for read in analysed_reads:
        name = read["sample"]
        if name not in samples:
            samples[name] = []
        samples[name].append(read)

    # Build summary for each sample
    summary = {}
    for name, reads in samples.items():
        valid_reads = [r for r in reads if r["status"] == "valid"]
        invalid_reads = [r for r in reads if r["status"] == "invalid"]

        # Mismatch reads: valid reads with at least one mutation
        mismatch_reads = [r for r in valid_reads if len(r["mutations"]) > 0]

        # Mismatch percentage (None if no valid reads)
        if len(valid_reads) > 0:
            mismatch_pct = (len(mismatch_reads) / len(valid_reads)) * 100
        else:
            mismatch_pct = None

        # Total mismatch events (sum of all mutations across valid reads)
        mismatch_events = sum(len(r["mutations"]) for r in valid_reads)

        # Substitution counts (e.g. "A>G": count)
        substitution_counts = {}
        for r in valid_reads:
            for m in r["mutations"]:
                key = f"{m['reference_base']}>{m['read_base']}"
                substitution_counts[key] = substitution_counts.get(key, 0) + 1

        # Position counts (position: count)
        position_counts = {}
        for r in valid_reads:
            for m in r["mutations"]:
                pos = m["position"]
                position_counts[pos] = position_counts.get(pos, 0) + 1

        # Top 3 positions (ties broken by earlier position)
        sorted_positions = sorted(
            position_counts.items(),
            key=lambda x: (-x[1], x[0])   # highest count first, then lowest position
        )
        top_positions = [pos for pos, count in sorted_positions[:3]]

        summary[name] = {
            "group": reads[0]["group"],
            "valid_reads": len(valid_reads),
            "invalid_reads": len(invalid_reads),
            "mismatch_reads": len(mismatch_reads),
            "mismatch_percentage": mismatch_pct,
            "mismatch_events": mismatch_events,
            "substitution_counts": substitution_counts,
            "position_counts": position_counts,
            "top_positions": top_positions,
        }

    # Print summary grouped by group
    _print_summary(summary)

    return summary


def _print_summary(summary):
    """Print a formatted summary table for each experimental group."""
    # Group samples by their group label
    groups = {}
    for name, data in summary.items():
        g = data["group"]
        if g not in groups:
            groups[g] = []
        groups[g].append(name)

    for group_name in sorted(groups.keys()):
        sample_names = sorted(groups[group_name])
        print(f"\n{'='*60}")
        print(f"GROUP: {group_name.upper()}")
        print(f"{'='*60}")

        # Table 1: per-sample summary
        print(f"\n{'Sample':<12} {'Valid':>6} {'Invalid':>8} {'%Mismatch':>10} "
              f"{'Events':>7} {'Top positions'}")
        print("-" * 70)
        for name in sample_names:
            d = summary[name]
            pct_str = f"{d['mismatch_percentage']:.1f}" if d['mismatch_percentage'] is not None else "N/A"
            top_str = ", ".join(str(p) for p in d["top_positions"]) if d["top_positions"] else "none"
            print(f"{name:<12} {d['valid_reads']:>6} {d['invalid_reads']:>8} "
                  f"{pct_str:>10} {d['mismatch_events']:>7} {top_str}")

        # Table 2: substitution mutations
        # Collect all substitution types across all samples in this group
        all_subs = set()
        for name in sample_names:
            all_subs.update(summary[name]["substitution_counts"].keys())
        all_subs = sorted(all_subs)

        if all_subs:
            print(f"\nSubstitution mutations (count, % per 100 valid reads):")
            header = f"{'Sub':<8}"
            for name in sample_names:
                header += f" {name:>18}"
            print(header)
            print("-" * (8 + 19 * len(sample_names)))

            for sub in all_subs:
                row = f"{sub:<8}"
                for name in sample_names:
                    d = summary[name]
                    count = d["substitution_counts"].get(sub, 0)
                    if d["valid_reads"] > 0:
                        pct = (count / d["valid_reads"]) * 100
                        row += f" {count:>5} ({pct:>5.1f}%)"
                    else:
                        row += f" {count:>5} (  N/A)"
                print(row)
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.

- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python

    {

        "treated_mean": 2.0,

    }
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

### 💻 Enter your code here

````python
def compare_positions(sample_summary, control_group="control", treated_group="treatment"):
    """Compare top mismatch positions between control and treated groups.

    Args:
        sample_summary: Dict as returned by summarise_samples().
        control_group: Label for the control group (default "control").
        treated_group: Label for the treated group (default "treatment").

    Returns:
        A list of dicts, one per top position, with keys:
            'position', 'control_n', 'treated_n',
            'control_mean', 'treated_mean', 'difference',
            'statistic', 'p_value'.
        Returns [] if no positions found.
    """
    # Step 1: Combine position_counts across all samples in both groups
    combined_counts = {}
    for name, data in sample_summary.items():
        if data["group"] in (control_group, treated_group):
            for pos, count in data["position_counts"].items():
                combined_counts[pos] = combined_counts.get(pos, 0) + count

    if not combined_counts:
        return []

    # Step 2: Select top 3 positions (highest count, ties broken by earlier position)
    sorted_positions = sorted(
        combined_counts.items(),
        key=lambda x: (-x[1], x[0])
    )
    top_positions = [pos for pos, count in sorted_positions[:3]]

    # Step 3: For each top position, calculate per-sample percentages
    results = []
    for pos in top_positions:
        control_pcts = []
        treated_pcts = []
        control_n = 0
        treated_n = 0

        for name, data in sample_summary.items():
            # Skip samples with no valid reads
            if data["valid_reads"] == 0:
                continue

            # Mismatch count at this position for this sample
            mismatch_at_pos = data["position_counts"].get(pos, 0)
            pct = (mismatch_at_pos / data["valid_reads"]) * 100

            if data["group"] == control_group:
                control_pcts.append(pct)
                control_n += 1
            elif data["group"] == treated_group:
                treated_pcts.append(pct)
                treated_n += 1

        # Calculate means
        control_mean = sum(control_pcts) / len(control_pcts) if control_pcts else None
        treated_mean = sum(treated_pcts) / len(treated_pcts) if treated_pcts else None

        # Difference
        if control_mean is not None and treated_mean is not None:
            difference = treated_mean - control_mean
        else:
            difference = None

        # Statistical test
        if control_pcts and treated_pcts:
            stat_result = calculate_statistics(control_pcts, treated_pcts)
            if stat_result is not None:
                statistic = stat_result["statistic"]
                p_value = stat_result["p_value"]
            else:
                statistic = None
                p_value = None
        else:
            statistic = None
            p_value = None

        results.append({
            "position": pos,
            "control_n": control_n,
            "treated_n": treated_n,
            "control_mean": control_mean,
            "treated_mean": treated_mean,
            "difference": difference,
            "statistic": statistic,
            "p_value": p_value,
        })

        # Print one line per position
        if p_value is not None:
            print(f"Position {pos}: control mean = {control_mean:.2f}%, "
                  f"treated mean = {treated_mean:.2f}%, "
                  f"diff = {difference:+.2f} pp, "
                  f"n={control_n}/{treated_n}, p={p_value:.4f}")
        else:
            print(f"Position {pos}: comparison could not be calculated")

    return results
````
